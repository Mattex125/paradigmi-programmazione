## Esercizio 1 - `mirror`

Scrivi `mirror`, che scambia ricorsivamente il sottoalbero sinistro e quello destro in ogni nodo.
Quale proprieta' deve avere `mirror (mirror t)`?

```ocaml
(*deve avere la proprietà di convoluzione *)

let rec mirror (t : 'a tree) : 'a tree =
  match t with
  | Leaf -> Leaf
  | Node (l,m,r)-> Node (mirror r, m, mirror l)
  
```

```ocaml
let t = Node (Node (Leaf, 1, Leaf), 2, Node (Leaf, 3, Leaf));;
assert (mirror Leaf = Leaf);;
assert (mirror t = Node (Node (Leaf, 3, Leaf), 2, Node (Leaf, 1, Leaf)));;
assert (mirror (mirror t) = t);;
Printf.printf "Test mirror: superato\n";;
```

## Esercizio 2 - `member`

Scrivi `member`, che controlla se un elemento compare nell'albero.
Non assumere che l'albero sia ordinato.

```ocaml
let rec member (x : 'a) (t : 'a tree) : bool =
  match t with
  | Leaf -> false
  | Node (l,m,r) -> if m=x || member x l || member x r then true else false
```

```ocaml
let t = Node (Node (Leaf, 1, Leaf), 2, Node (Leaf, 3, Leaf));;
assert (member 2 t);;
assert (member 3 t);;
assert (not (member 4 t));;
assert (not (member 1 Leaf));;
Printf.printf "Test member: superato\n";;
```

## Esercizio 3 - `leaves`

Scrivi `leaves`, che conta quante foglie contiene l'albero.
Verifica anche la relazione tra `leaves t` e la dimensione dell'albero.

```ocaml
let rec leaves (t : 'a tree) : int =
  let rec go (t : 'a tree) (acc : int) : int =
    match t with
    | Leaf -> acc + 1
    | Node (l, _, r) -> go l (go r acc)
  in go t 0
```

```ocaml
let t = Node (Node (Leaf, 1, Leaf), 2, Node (Leaf, 3, Leaf));;
assert (leaves Leaf = 1);;
assert (leaves t = 4);;
assert (leaves (Node (Leaf, 0, Leaf)) = 2);;
Printf.printf "Test leaves: superato\n";;
```
