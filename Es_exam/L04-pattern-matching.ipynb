## Esercizio 1 - `ultimo`

Scrivi `ultimo`, che restituisce l'ultimo elemento della lista dentro `Some`.
Se la lista e' vuota deve restituire `None`.

Usa il pattern matching e ricorri sulla parte piu' piccola della lista.

```ocaml
let rec ultimo (l : 'a list) : 'a option =
  match l with
  | [] -> None
  | [x] -> Some x 
  | prev::x -> ultimo x
```

### Test di `ultimo`

Esegui questa cella dopo aver completato la funzione.

```ocaml
assert (ultimo [] = None);;
assert (ultimo [ 7 ] = Some 7);;
assert (ultimo [ 1; 2; 3; 4 ] = Some 4);;
```

## Esercizio 2 - `pari`

Scrivi `pari`, che restituisce una nuova lista con i soli elementi pari,
nello stesso ordine in cui compaiono nella lista di partenza.

Gestisci esplicitamente il caso della lista vuota.

```ocaml
let rec pari (l : int list) : int list =
  match l with
    | [] -> []
    | this :: rest -> if (this mod 2)=0 then List.append [this] (pari rest) else pari rest
```

### Test di `pari`

Esegui questa cella dopo aver completato la funzione.

```ocaml
assert (pari [] = []);;
assert (pari [ 1; 3; 5 ] = []);;
assert (pari [ 1; 2; 3; 4; 6 ] = [ 2; 4; 6 ]);;
assert (pari [ -4; -1; 0; 7 ] = [ -4; 0 ]);;
Printf.printf "Test pari: superato\n";;
```
