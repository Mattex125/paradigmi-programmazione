## Esercizio 1 - usare `fold`

Riscrivi con `fold`, senza usare `rec`, le funzioni `maximum`, `reverse` e `member`.
Per `maximum`, scegli con attenzione il tipo del risultato.

```ocaml
(* Funzione fornita dalla traccia. *)
let rec fold combine initial = function
  | [] -> initial
  | x :: resto -> combine x (fold combine initial resto);;

(* Firma di fold:
  ('a -> 'b -> 'b) -> 'b -> 'a list -> 'b
*)
let maximum (l : int list) : int option =
  match l with
  | [] ->
      (* Una lista vuota non ha un massimo. *)
      None
  | this :: rest ->
      (* Il primo elemento diventa il massimo iniziale. *)
      fold
        (fun elemento massimo_corrente ->
          (* fold passa prima l'elemento e poi il risultato accumulato. *)
          match massimo_corrente with
          | None ->
              (* Caso teorico: non c'e' ancora un massimo. *)
              Some elemento
          | Some massimo ->
              (* Conserva il maggiore tra il nuovo elemento e il massimo. *)
              Some (max elemento massimo))
        (Some this)
        (* Il fold controlla tutti gli elementi rimanenti. *)
        rest;;

(*let reverse (l : 'a list) : 'a list =
  failwith "da completare";;
*)
let rec member (x : 'a) (t : 'a tree) : bool =
  match t with 
  | Leaf -> false
  | Node (l,m,r)-> if m=x then true else member x l || member x r
```

```ocaml
let t = Node (Node (Leaf, 1, Leaf), 2, Node (Leaf, 3, Leaf));;
assert (maximum [] = None);;
assert (maximum [ 3; 1; 8; 2 ] = Some 8);;
assert (maximum [ -4; -2; -9 ] = Some (-2));;
(*assert (reverse [] = []);;
//assert (reverse [ 1; 2; 3 ] = [ 3; 2; 1 ]);;
assert (reverse [ "a"; "b" ] = [ "b"; "a" ]);;
*)
assert (member 2 t);;
assert (not (member 9 t));;
Printf.printf "Test maximum, reverse, member: superato\n";;
```

## Esercizio 2 - `fold_tree`

Estrai la forma comune delle funzioni sugli alberi definendo `fold_tree`.
Decidi quanti argomenti servono e quale parte della funzione deve decidere il chiamante.

```ocaml
let rec fold_tree on_leaf on_node (t : 'a tree) =
  match t with
  | Leaf -> on_leaf
  | Node (l, m, r) ->
      let sinistro = fold_tree on_leaf on_node l in
      let destro = fold_tree on_leaf on_node r in
      on_node sinistro m destro
```

```ocaml
let t = Node (Node (Leaf, 1, Leaf), 2, Node (Leaf, 3, Leaf));;
let somma = fold_tree 0 (fun sinistro valore destro -> sinistro + valore + destro);;
let conta_nodi = fold_tree 0 (fun sinistro _ destro -> 1 + sinistro + destro);;
assert (somma t = 6);;
assert (conta_nodi t = 3);;
assert (somma Leaf = 0);;
Printf.printf "Test fold_tree: superato\n";;
```
